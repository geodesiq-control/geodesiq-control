```python
import matplotlib.pyplot as plt
import numpy as np

from geodesiq import ControlModel, about
```

```python
def H_fun(x, z):
    return np.array([[z, x], [x, -z]])


def H_partial(x, z):
    return np.array([[1, 0], [0, -1]])


alpha = 50
beta = 50
x = 1
z0 = -10
zf = -z0

model = ControlModel(H_fun, H_partial)
model.set_parameters(x=x)
model.set_control(control_name="z", pulse_initial=z0, pulse_final=zf, initial_state=0, alpha=alpha, beta=beta)
model.solve_problem()
```

```python
tau, pulse = model.tau, model.control_sol

fig, ax = plt.subplots()
ax.plot(tau, pulse)

ax.axhline(z0, c="r", ls="--")
ax.axhline(zf, c="r", ls="--")

ax.set_xlabel(r"$\tau$")
ax.set_ylabel(r"$z(\tau)$");
```
